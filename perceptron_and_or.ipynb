Perceptron (Binary Classification)

```python
import numpy as np
```

Example : Study hours->Pass/Fail

```python
# Training Data
x=np.array([1,2,3,4])
# Study hours
y=np.array([0,0,1,1])
# Pass(1)/Fail(0)
x,y
```

```python
# Initialize parameters
w=0.0
b=0.0
learning_rate=1
epochs=5
print("Initail weight:",w)
print("Initial Bias",b)
```

Define Activation Function(Step function)

```python
def step_function(z):
  return 1 if z>=0 else 0
```

Training Loop

```python
for epoch in range(epochs):
  print(f"\nEpoch {epoch+1}")
  for i in range(len(x)):
    # 1.Compute weightes sum
    z=w * x[i]+b
    # 2.Apply Activation function
    y_pred=step_function(z)
    # 3.Compute Error
    error=y[i]-y_pred
    # 4.Update weight
    w=w+learning_rate * error *x[i]
    # 5.Update Bias
    b=b+learning_rate * error

    print(f"Input :{x[i]},Predicted value :{y_pred},Actual :{y[i]},Weight(w) :{w},Bias(b) :{b}")
```

Testing the model

```python
test_value=2.9
z=w*test_value+b
prediction=step_function(z)
print(f"For study hours : {test_value},Prediction :",prediction)
```

Perceptron AND Gate

```python
x=np.array([
    [0,0],
    [1,0],
    [0,1],
    [1,1]
])
y=np.array([0,0,0,1])
x,y
```

```python
# Initialize parameters
w=np.zeros(x.shape[1])
b=0
learning_rate=0.1
epochs=10

print("Initial weight : ",w)
print("Initial Bias : ",b)
```

```python
def step_function(z):
  return 1 if z>=0 else 0
```

```python
for epoch in range(epochs):
  print(f"\nEpoch {epoch+1}")
  for i in range(len(x)):
    z=np.dot(w,x[i])+b
    y_pred=step_function(z)
    error=y[i]-y_pred
    w=w+learning_rate * error * x[i]
    b=b+learning_rate * error
    print(f"\nInput :{x[i]},Predicted value :{y_pred},Actual :{y[i]},Weight(w) :{w},Bias(b) :{b}")
```

```python
print("Final Weight : ",w)
print("Final Bias : ",b)
```

```python
# Testing model
print("\nTesting AND Gate")
for i in range(len(x)):
  z=np.dot(w,x[i])+b
  prediction=step_function(z)
  print("Input : ",x[i],"Prediction : ",prediction)
```

```python
import matplotlib.pyplot as plt
```

A decision boundary is the dividing line (or surface, in higher dimensions) that separates different classes predicted by a classifier.
- For a perceptron, the boundary is linear(straight line)


```python
# Plotting the decision boundary
plt.scatter(x[:,0], x[:,1], c=y, cmap='bwr', edgecolors='k')

# Equation of line: w1*x1 + w2*x2 + b = 0 → x2 = -(w1*x1 + b)/w2
x_vals = np.linspace(-0.5, 1.5, 100)
y_vals = -(w[0]*x_vals + b)/w[1]
plt.plot(x_vals, y_vals, 'g--', label='Decision Boundary')

plt.title("Perceptron Decision Boundary (AND Gate)")
plt.xlabel("x1")
plt.ylabel("x2")
plt.legend()
plt.grid(True)
plt.show()
```
