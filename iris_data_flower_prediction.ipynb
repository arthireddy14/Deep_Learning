Goal:
Predict which flower species an iris belongs to.
Classes:

Setosa Versicolor Virginica

Input layer(4 neurons)

Hidden layer(10 neurons,ReLU)

Hidden Layer(5 neurons,ReLU)

Output Layer(3 neurons,Softmax)

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow import argmax
# converts probability output->class index
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
# load iris dataset
iris=load_iris()
x=iris.data
y=iris.target
target_names=iris.target_names
print(target_names)
```

```python
# split the dataset
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
# define model architecture
model=Sequential()
model.add(Dense(10,activation='relu',input_shape=(x_train.shape[1],)))
model.add(Dense(5,activation='relu'))
model.add(Dense(3,activation='softmax'))
model.summary()
```

Compile the model

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history=model.fit(x_train_scaled,y_train,epochs=50,validation_split=0.1,verbose=1)
```

```python
test_loss,test_accuracy=model.evaluate(x_test_scaled,y_test)
print(f"Test accuracy : ",test_accuracy)
```

```python
# make predictions
y_pred_proba=model.predict(x_test_scaled)
y_pred=argmax(y_pred_proba,axis=1)
# Argmax selects the highest probability. [0.98,0.01,0.01]->class 0

# calaculate accuracy
accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy : {accuracy}")
```
