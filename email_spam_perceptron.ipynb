🏢 Business Scenario: Email Spam Detection for a Startup
A startup company is building a basic email filtering system to classify emails as:
1 → Spam
0 → Not Spam
The company wants to first implement a simple Perceptron model before moving to complex neural networks.
They identified two key binary features from emails:
Contains Suspicious Keywords (e.g., “lottery”, “prize”, “urgent”)
Yes → 1
No → 0
Sender is Unknown
Yes → 1
No → 0
So each email is represented as:

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Features[Suspicious keywords,Unknown Sender]
x=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y=np.array([0,0,0,1])
# Email is spam if both features are true
```

```python
# Assume trained weights
w = np.array([0.5, 0.5])
b = -0.7
```

```python
# Plot points
plt.scatter(x[:,0], x[:,1], c=y, cmap='bwr', edgecolors='k')

# Decision boundary line
x_vals = np.linspace(-0.5, 1.5, 100)
y_vals = -(w[0]*x_vals + b)/w[1]
plt.plot(x_vals, y_vals, 'g--', label='Decision Boundary')

plt.title("Perceptron Decision Boundary (Spam Detection)")
plt.xlabel("Suspicious Keywords")
plt.ylabel("Unknown Sender")
plt.legend()
plt.grid(True)
plt.show()
```

- What happens if we change E3 output to 0?
Suppose the email with (0,1) (unknown sender but no suspicious keywords) is labeled as not spam (0) instead of spam.
- Will perceptron still converge?
Yes, because the dataset remains linearly separable: only (1,1) is spam, all others are not spam. The perceptron can still find a line that separates (1,1) from the rest.
- But if we changed it differently (e.g., made (1,0) or (0,1) spam too):
The dataset would resemble an XOR pattern, which is not linearly separable. In that case, the perceptron would fail to converge.

Limitation exposed:
A single-layer perceptron can only solve linearly separable problems.

It cannot learn XOR-like patterns.

This limitation is why multi-layer neural networks (with hidden layers) are needed for more complex decision boundaries.
