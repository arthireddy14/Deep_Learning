XOR Problem with Multi Layer Neural Network(Forward Propagation)

Architecture:

2 Inputs->2 Hidden Neurons ->1 Output

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# input xor table
x=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y=np.array([0,1,1,0])
x,y
```

Initialize Weights and  Bias(Manually Choosen)

```python
# hidden layer weights and biases

# w=np.array([1,1,1,1,1,-2])
# b=np.array([-0.5,-1.5,-0.5])

w1,w2,b1=1,1,-0.5
w3,w4,b2=1,1,-1.5

# output layer weights and bias
w5,w6,b3=1,-2,-0.5
print("Weights Initialized Successfully")
```

Define Step Activation Function

```python
def step_function(z):
  return 1 if z>=0 else 0
```

Define Forward Propagation Function

```python
def forward_propagation(x1,x2):
  # hidden layer calculations
  h1=w1*x1+w2*x2+b1
  a_h1=step_function(h1)
  h2=w3*x1+w4*x2+b2
  a_h2=step_function(h2)
  # output layer calculations
  y=w5*a_h1+w6*a_h2+b3
  output=step_function(y)

  return output
```

Test Forward Propagation for All XOR Inputs

```python
print('XOR Results using ANN Forward Propagation')

for x1 in x:
  prediction=forward_propagation(x1[0],x1[1])
  print(f"Input :{x1} Output: {prediction}")
```

```python
# Plot decision boundary
x_min, x_max = x[:, 0].min() - 1, x[:, 0].max() + 1
y_min, y_max = x[:, 1].min() - 1, x[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))

# Predict probabilities for each point in the grid
z = np.array([forward_propagation(i,j) for i,j in zip(xx.ravel(),yy.ravel())])
z = z.reshape(xx.shape)

plt.contourf(xx, yy, z, levels=[0, 0.5, 1], alpha=0.3, colors=["blue", "red"])
plt.scatter(x[:, 0], x[:, 1], c=y, edgecolors="k", cmap=plt.cm.RdBu)
plt.title("Decision Boundary (Logistic Regression)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
